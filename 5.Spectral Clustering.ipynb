```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
# Restore the features for all 200 customers

X = dataset.iloc[:, [3, 4]].values
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Shape:", X_scaled.shape)
print("First five scaled rows:")
print(X_scaled[:5])
```

```python
from sklearn.cluster import SpectralClustering

sc = SpectralClustering(
    n_clusters=3,
    affinity="rbf",
    gamma=1.0,
    assign_labels="discretize",
    random_state=0
)

y_sc = sc.fit_predict(X_scaled)
```

```python
print("Cluster labels:")
print(y_sc)

print("\nNumber of clusters:", len(np.unique(y_sc)))
```

```python
dataset["Spectral_Cluster"] = y_sc

print("Customers:", len(dataset))
print("Cluster labels:", len(y_sc))

display(dataset.head(10))
```

```python
summary = dataset.groupby("Spectral_Cluster").agg(
    Customers=("CustomerID", "count"),
    Average_Income=("Annual Income (k$)", "mean"),
    Average_Spending=("Spending Score (1-100)", "mean")
).round(2)

display(summary)
```

```python
dataset.to_csv(
    "Spectral_Customer_Results.csv",
    index=False
)

summary.to_csv("Spectral_Cluster_Summary.csv")

print("Customer results and cluster summary saved.")
```
