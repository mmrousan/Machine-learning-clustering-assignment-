```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
# Restore the features for all 200 customers

X = dataset.iloc[:, [3, 4]].values
```

```python
X[:5]
```

```python
from sklearn.cluster import AffinityPropagation

# Create and train the clustering model

aff = AffinityPropagation(random_state=5)
y_aff=aff.fit_predict(X)
```

```python
# Verify that the lengths match
print("Customers:", len(dataset))
print("Cluster labels:", len(y_aff))
```

```python
# Add each customer's cluster to the dataset
dataset["Cluster"] = y_aff
```

```python
print("Number of clusters:", len(set(y_aff)))
display(dataset.head(10))
```

```python
summary = dataset.groupby("Cluster").agg(
    Customers=("CustomerID", "count"),
    Average_Income=("Annual Income (k$)", "mean"),
    Average_Spending=("Spending Score (1-100)", "mean")
).round(2)

display(summary)
```
